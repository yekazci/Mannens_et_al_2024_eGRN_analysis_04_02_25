From Epiregulon vignette:

Epiregulon estimates the regulatory potential using one of the three measures:
1) correlation between TG and TF or between TG and the product of TF and RE,
2) mutual information between TG and TF expression or between TG and the product of TF and RE, 
3) Wilcoxon test statistics of target gene expression in cells jointly expressing all 3 elements vs cells that do not.

=================================================================

From the method above, one can appreciate that Pando finds overlaps\
between input genomic ranges of peak regions in the peaks count feature\
names and the "regions" if provided by the user. Then, it stores the\
matching peak ranges' positional indices (in the peak count matrix)\
in the newly created 'regions' object. This 'regions' object stores\
the peak region raw indices in the 'peaks' slot. The new peak regions'\
genomic ranges are stored in the 'ranges' slot of the Refions object.

When infer_eGRN function is run, Pando will take the region names from\
the Regions object and also retrieve the paek counts for each regions by\
indexing the peak counts matrix in the Seurat object using the peaks indices\
in the Region object.

```r
# In this example, we apply Wilcoxon test on cell aggregates of 10 cells. 
# We use the Wilcoxon weight method  because we are interested in computing 
# cell type-specific weights.

set.seed(1010)

Pando_regulon_df_weighted <- epiregulon::addWeights(regulon = Pando_regulon_df,
                                expMatrix  = GeneExpressionMatrix,
                                exp_assay  = "logcounts",
                                peakMatrix = PeakMatrix_PANDO,
                                peak_assay = "counts",
                                clusters = GeneExpressionMatrix$main_cell_types_renamed,
                                aggregateCells = TRUE,
                                min_targets = 10, # Calc. for TFs with minimum num. of targets. Default is 10. 
                                method = "wilcox",
                                useDim = "PCA")
```
