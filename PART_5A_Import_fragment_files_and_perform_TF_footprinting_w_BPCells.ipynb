```r
# Function to apply prefix to cell barcodes
add_prefix_to_cells <- function(fragment_obj, prefix) {
  prefix_cell_names(fragment_obj, paste0(prefix, ":"))
}
```

In this notebook, I firstly stored all of the fragment files from all 10X scATAC-seq output files on disk\
using BPCells functions. Then, form the metadata of mannens et al seurat object, I identified the 10X\
identity names of those cell barcodes being included in my seurat object. I realized that my cell barcodes\
are named in follwing format: 10X365_2:ACCAATATCAATGACC. Then, I found the fragment folders that are named\
with the 10X identity prefixes preceding the cell barcodes. I merged those fragment objects into a single\
fragment object. Then, I subsetted those fragment objects into cells that are included in my seurat object.\
Finally, I wrote this final fragment object into a new directory. Now, it is ready to be used.\
I can also save it on memory and use in other workflows such as Signac.
