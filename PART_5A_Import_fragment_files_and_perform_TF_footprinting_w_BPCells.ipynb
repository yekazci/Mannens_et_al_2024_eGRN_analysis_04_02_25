```r
# Function to process each fragment file
process_fragment <- function(file) {
  # Create a clean output directory name
  new_file_name <- basename(file) |> 
                      str_remove_all(pattern = '\\.') |> 
                      str_remove_all(pattern = 'tsvgz')
  
  out_dir <- here::here('bpcells_counts', 'bpcells_fragments', paste0(new_file_name, "_bpcells_dir"))

  # Import and write fragments
  if (!file.exists(out_dir)) {
    frags <- open_fragments_10x(file) |> 
      write_fragments_dir(out_dir)
  } else {
    frags <- open_fragments_dir(out_dir)
  }

  return(frags)
}

# Process all files
fragment_objects <- lapply(fragment_files, process_fragment)

# Merge all fragment objects
merged_fragments <- do.call(c, fragment_objects)

# Optionally, write merged fragments to disk
write_fragments_dir(merged_fragments, "bpcell_merged_fragments_dir")

# Print merged object info
print(merged_fragments)
```
