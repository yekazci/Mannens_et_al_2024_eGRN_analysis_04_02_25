If we use AUCell to calculate TF activity scores instead of weighted mean method (default) in epiregulon package,\
we can also calculate regulon specificity scores (RSS) from AUC scores using the following method from PyScenic:

```r
################## PYTHON CODE FROM PYSCENIC ###########################################

# rom math import sqrt

# import numpy as np
# import pandas as pd
# from scipy.spatial.distance import jensenshannon


# def regulon_specificity_scores(auc_mtx, cell_type_series):
#     """
#     Calculates the Regulon Specificty Scores (RSS). [doi: 10.1016/j.celrep.2018.10.045]

#     :param auc_mtx: The dataframe with the AUC values for all cells and regulons (n_cells x n_regulons).
#     :param cell_type_series: A pandas Series object with cell identifiers as index and cell type labels as values.
#     :return: A pandas dataframe with the RSS values (cell type x regulon).
#     """

#     cell_types = list(cell_type_series.unique())
#     n_types = len(cell_types)
#     regulons = list(auc_mtx.columns)
#     n_regulons = len(regulons)
#     rss_values = np.empty(shape=(n_types, n_regulons), dtype=np.float32)

#     def rss(aucs, labels):
#         # jensenshannon function provides distance which is the sqrt of the JS divergence.
#         return 1.0 - jensenshannon(aucs / aucs.sum(), labels / labels.sum())

#     for cidx, regulon_name in enumerate(regulons):
#         for ridx, cell_type in enumerate(cell_types):
#             rss_values[ridx, cidx] = rss(
#                 auc_mtx[regulon_name], (cell_type_series == cell_type).astype(int)
#             )

#     return pd.DataFrame(data=rss_values, index=cell_types, columns=regulons)
```

```r
########################### R VERSION  ################################################################

# To calculate regulon specificity scores from AUC scores as in Pyscenic codebase:

library(philentropy)  # For Jensen-Shannon divergence
library(dplyr)        # For data manipulation

# Function to compute Regulon Specificity Scores (RSS)
regulon_specificity_scores <- function(auc_mtx, cell_type_vector) {
  
  # Get the unique cell types from the input vector
  cell_types <- unique(cell_type_vector)
  
  # Get the column names of the AUC matrix (i.e., regulon names)
  regulons <- colnames(auc_mtx)
  
  # Initialize an empty matrix to store RSS values
  # Rows correspond to cell types, columns correspond to regulons
  rss_values <- matrix(NA, nrow = length(cell_types), ncol = length(regulons),
                       dimnames = list(cell_types, regulons))

  # Function to compute the RSS for a given regulon and cell type
  compute_rss <- function(auc_values, labels) {
    auc_dist <- auc_values / sum(auc_values)  # Normalize AUC values to get a probability distribution
    label_dist <- labels / sum(labels)        # Normalize binary labels to get a probability distribution
    js_distance <- JSD(rbind(auc_dist, label_dist))  # Compute Jensen-Shannon divergence
    return(1.0 - sqrt(js_distance))  # Convert JS distance to RSS score (higher means more specific)
  }

  # Loop over each regulon (column of auc_mtx)
  for (cidx in seq_along(regulons)) {
    # Loop over each cell type
    for (ridx in seq_along(cell_types)) {
      # Create a binary mask: 1 for cells of the current type, 0 otherwise
      cell_mask <- as.integer(cell_type_vector == cell_types[ridx])  
      
      # Compute RSS for the current regulon and cell type
      rss_values[ridx, cidx] <- compute_rss(auc_mtx[, cidx], cell_mask)
    }
  }
  
  # Convert the RSS matrix to a data frame before returning
  return(as.data.frame(rss_values, row.names = cell_types))
}
```

```r
library(ggplot2)
library(dplyr)
library(scales)  # For controlling breaks on y-axis

plot_rss <- function(rss_df, cell_type, top_n = 5, max_n = NULL) {
  
  # Extract the RSS values for the specified cell type
  data <- rss_df[cell_type, , drop = FALSE] %>% t() %>% as.data.frame()
  colnames(data) <- "RSS"
  data$Regulon <- rownames(data)
  
  # Sort the data by RSS values in descending order
  data <- data %>% arrange(desc(RSS))
  
  # Set the maximum number of regulons to display
  if (is.null(max_n)) {
    max_n <- nrow(data)
  }
  
  # Select the top `max_n` regulons
  data <- data[1:max_n, ]
  
  # Create the base plot
  p <- ggplot(data, aes(x = seq_along(Regulon), y = RSS)) +
    geom_point() +  # Plot RSS values as points
    ylim(floor(min(data$RSS) * 100) / 100, ceiling(max(data$RSS) * 100) / 100) +  # Adjust y-axis
    labs(
      x = "Regulon",
      y = "RSS",
      title = paste("Regulon Specificity Scores for", cell_type)
    ) +
    theme_minimal() +
    theme(axis.text.x = element_blank())  # Hide x-axis labels
  
  # Add labels for the top `top_n` regulons
  top_labels <- data[1:top_n, ]
  p <- p + 
    geom_point(data = top_labels, aes(x = seq_along(Regulon)[1:top_n], y = RSS), color = "red") +
    geom_text(
      data = top_labels,
      aes(x = seq_along(Regulon)[1:top_n] + (max_n / 25), y = RSS, label = Regulon),
      color = "red",
      size = 3,
      hjust = 0
    )
  
  return(p)
}
```

IMPORTANT NOTE:

When TF activity scores got calculated by weighted mean approach, the wilcoxon\
test statistic values are z-score transformed and stored as 'weights'. It contains\
both negative and positive values reflecting respectively the repressory and\
activatory regulatory directions. However, regulon specificity score is calculated\
by jensen-shannon divergence method and expects non-negative input values that will\
be converted into probability distributions and then their distances will be calculated.

As can be seen below, when directly used JS distance matrix calculation on negative\
values, the result is full of NaNs !!!!!!!!!!!!!!!!

There is a problem. When I calculate the TF activity scores using the default method,\
"weighted mean", the scores can be both negative and postive values.\
However, RSS score calculates a probability distribution and so the TF activity scores\
need to be transformed.

Actually, triggered by this issue, I need to have a more in-depth understanding of the\
weight values, how to interprete the negative and positive ones, or do I actually\
need to keep negative ones etc.

===========================

There are different options to address negative values:

1. One can discard them completely before calculating JS divergence distances.
2. One can set the negative values to zero if activatory regulatory directions are only desired.
3. One can take absolute of all weights and calculate JS distance metrics. However, during plotting,\
   activatory and repressory TFs should be labelled in different colurs to distinguish them.
4. One can create respective matrices for negative and positive weights and perform the JS divergence for both.

Below, I will go with the second option using pmax function, which provides utility to replace all negative values\
in an input matrix with user-given integer.

```r
positive_mtx <- pmax(score_TF_activity_by_clusters, 0)  # Replace negative values with 0
```

AUCELL does not use weight information. It looks for enrichment of target genes of TFs\
in the ranked genes within each cell. It is likely that the those high AUC score having\
TFs are activatory but there is not explicit info about this.

When weighted mean is used in calculation of TF activity scores, weights in regulon\
data frame are used. Calculated scores are negative signed if TF is repressor and\
positive if TF is activator. However, to calculate RSS scores, we need non-negative\
values. THere are different options to transform TF activity scores. One is to\
disregard negative values by replacing them with zeros. Then calculate RSS.

One can also take the absolute of each score and calculate RSS. However, to differentiate\
activatory and repressory TFs, we need to retrieve the signs of their respective weights\
in the weighted regulon object and use it to differently colour the activatory and\
repressory TFs in the plot. We can also add activatory and repressory categorical\
variables to the relevant objects.

====================================================================

It is possible to aggregate the correlation scores of same TF-TG pairs\
in epiregulon while computing regulons. THere is a parameter for this,
in the getRegulon function. It might be better to use this option\
to compare Epiregulon results with Pando results, as Pando aggregates\
as default option. However, It might be also possible to go without\
aggregation in Pando. It does aggregation during find_modules step\
so there might be a parameter to unset it.

=====================================================================

So I should keep in mind that calculated weights refer to repressory or\
activatory regulation by TFs according to sign of the values, and\
this is used in TF activity calculations as long as AUCELL option\
