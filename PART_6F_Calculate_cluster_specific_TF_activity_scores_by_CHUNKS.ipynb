About output of the function when argumented wit cluster info:

Yep, that makes sense! Even when you provide clusters, \
the function still outputs a TF-by-cell barcode matrix—it\
just adjusts calculations per cluster internally.

What changes with clusters is how activity scores are computed:

Without clusters: The entire dataset is treated as one group.\
With clusters: It normalizes within each cluster, but still \
outputs the same type of matrix (TFs as rows, cell barcodes as columns).\
If you were expecting a TF-by-cluster summary, you might need to manually\
aggregate the scores by cluster (e.g., taking the mean per cluster). Would that help? 🚀

===========================

About the "calculating frequency" message that is printed when cluster info provided:


The message "calculating frequency" appears when you provide the \
clusters argument because the function computes the frequency of\
transcription factor (TF)-target interactions within each cluster.

Here's why:

When clusters are provided: The function calculates cluster-specific\
frequencies to normalize the activity scores for each cell within \
its respective cluster. This step helps adjust for the number of \
TF-target interactions specific to that cluster.

Message behavior: The message "calculating frequency..." is triggered \
when it's performing this frequency calculation for each cluster.

When clusters are not provided: The function skips cluster-specific \
calculations and simply calculates overall frequencies for the whole dataset.\
It doesn't need to calculate frequency per cluster, so the message is not printed.

In short:

With clusters: Frequency is calculated at the cluster level.
Without clusters: Frequency is calculated globally, and \
no "calculating frequency" message is printed.
If you want to further explore the frequency calculations \
or prevent unnecessary messages, you could check how the \
calculateFrequency() function is used in the code or adjust \
logging verbosity. Let me know if you'd like help with that!

==========================================================
IMPORTANT NOTE:

epiregulon::calculateActivity function calculates activity for each TF\
individually and independently per each cell. Therefore, one can divide\
weighted regulon dataframe into chunks of TFs and calculate their respective\
activity scores and then combine the results at the end.

Importantly, setting "normalized" argument to FALSE (default) or TRUE does\
not induce any change in the calculated scores for TFs in chunks, since
it is not something done chunk-wise, as it resulted no discrepancy in the\
calculated values when I tested.

IMPORTANT NOTE 2 :

I will use "normalized = TRUE" because it seems more reasonable for my\
downstream goal, which is to compare TF activity levels between \
clusters. This is the explanation by chatGPT:

If your goal is to identify TF-gene networks that are \
more active in one cluster compared to the rest, using \
normalize = TRUE is generally recommended, but you should\
also consider additional statistical tests after computing\
TF activity scores.

Why normalize = TRUE? :
Removes global expression biases: Ensures that differences \
in TF activity are not simply due to overall gene expression levels.\
Compares TF activity across clusters more fairly: Since target gene\
expression levels may vary across clusters, normalizing \
by mean expression and frequency ensures you are detecting\
true differences in TF activity rather than artifacts.
