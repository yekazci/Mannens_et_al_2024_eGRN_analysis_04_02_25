```r
# Normalize by reads-per-cell
mat <- multiply_cols(mat = counts_RNA, 1/Matrix::colSums(counts_RNA))
```

```r
# Log normalization
mat <- log1p(mat * 10000) # Log normalization
```

```r
stats <- matrix_stats(mat, row_stats="variance")

# To keep the example small, we'll do a very naive variable gene selection
variable_genes <- order(stats$row_stats["variance",], decreasing=TRUE)  |>  
  head(1000)  |>  
  sort()

mat_norm <- mat[variable_genes,]
```

```r
mat_norm
```

===========================================================================

Since there will be some iterative computations in the PCA computation, \
we save the current matrix to a temporary directory so that the normalizations \
above will not be performed repeatedly in the pca computation by slowing down the\
process.

```r
mat_norm <- mat_norm |> write_matrix_dir(dir = here::here('temp/mat'))
```

```r
gene_means <- stats$row_stats["mean",variable_genes]
```

```r
gene_vars <- stats$row_stats["variance", variable_genes]
```

```r
mat_norm <- (mat_norm - gene_means) / gene_vars
```

```r
# Next we calculate the TF-IDF normalization. This formula is the TF-IDF variant from Stuart et al.

mat_lsi <- counts_ATAC  |> 
  multiply_cols(1 / Matrix::colSums(counts_ATAC))  |> 
  multiply_rows(1 / Matrix::rowMeans(counts_ATAC))

mat_lsi <- log1p(10000 * mat_lsi)
```

```r
mat_lsi
```

```r
# Just like for the RNA, we save our matrix before running PCA. 

mat_lsi <- mat_lsi |> write_matrix_dir(dir = here::here('temp/mat_lsi'))
```

```r
mat_lsi
```

```r
# Finally, we will do a z-score normalization on the LSI matrix then run PCA. 
# This is standard practice for running a PCA, but is not commonly done 
# on ATAC-seq datasets due to the fact that it greatly increases memory usage. 
# In other methods, the 1st PC is highly correlated to the number of reads 
# per cell, and is thrown out as an empirical correction.

# Luckily, BPCells can avoid this memory usage so we can just normalize our data 
# and run PCA as usual

# Compute colMean and colVariance in one pass: 

cell_peak_stats <- matrix_stats(mat_lsi, col_stats="variance")$col_stats
cell_means <- cell_peak_stats["mean",]
cell_vars <- cell_peak_stats["variance",]
mat_lsi_norm <- mat_lsi  |> 
  add_cols(-cell_means)  |> 
  multiply_cols(1 / cell_vars)
```

```r
svd_atac <- BPCells::svds(mat_lsi_norm, k=10)
pca_atac <- multiply_cols(svd_atac$v, svd_atac$d)
pca_atac[1:4,1:4]
```

```r
# Next we calculate our UMAP

set.seed(12341512)
umap_atac <- uwot::umap(pca_atac)
umap_atac[1:4,]
```

```r
cor_to_depth <- tibble::tibble(
    abs_cor_to_depth = as.numeric(abs(cor(Matrix::colSums(mat_lsi), pca_atac))),
    PC=seq_along(abs_cor_to_depth)
  )
```

```r
cor_to_depth |> head()
```

```r
# SInce BPCells apples z-score normalization to the LSI matrix, its first PC component does not correlate
# with the sequencing depth as high as the raw TF-DF method of Signac.
```
